### Encapsulation

```python
class Car: # class
    # instructore/ atttributes
    def __init__(self, name, model, year, color, wheel, engine, interior):
        self.name = name
        self.model = model
        self.year = year
        self.color = color
        self.wheel = wheel
        self.engine = engine
        self.interior = interior


    # methods
    def idle(self):
        print(f"{self.name} is idle")
        
    def start(self):
        print(f"{self.name} is starting")
    
    def stop(self):
        print(f"{self.name} is stop")
    
    def wheels(self):
        print(f"{self.name} have {self.wheel} wheels")
    
    def colors(self):
        print(f"the color is {self.color}")
        
    def engines(self):
        print(f" the engine is {self.engine}")

    def interiors(self):
        print(self.interior)
```

### Access modifiers
   - public  (hum class kai andar aur class kai bhr use kr skte hai kuch bhi )
   - private  ( sirf class klai andar hi use ho ga )

```python
class person:
    def __init__(self, name , height, age , weight, gender):
        # public
        self.name = name
        self.height =height

        # private 
        self._age = age
        self.weight= weight
        self.gender = gender
    def full_info(self):
        print(f""" your name is {self.name}  
                   your height is {self.height}
                   your  age is {self._age}
                   your weight is {self.weight}
                   your gender {self.gender}""")
        # methods 
    def get_age(self):  # get the age 
        print(self._age)

        
    def set_age(self, new_age):   # set the age
        if (self._age < 0) or (self._age>100):
            print("invalid age")
        else:
            self._age = new_age
    
        
```

### Inheritance

### parent class kai methods ya attrubutes ko child class mai use krna 

```python
class Animal: # parent class
    def __init__(self,name,age,color,breed,gender,Type,food):
        self.name = name
        self.age = age
        self.color = color
        self.breed = breed
        self.gender= gender
        self.Type = Type
        self.food = food
    def sleep(self):
        print("Animal so skta hai ")
    def walk(self):
        print("parent class ka")
    def show_color(self):
        print(f" Animal color {self.color}")
    def food(self):
        print(f" animal food {self.food}")
    def full_info(self):
        print(f"""
                    Welcome
                Animal name {self.name}
                Animal age {self.age}
                Animal color {self.color}
                Animal breed {self.breed}
                Animal gender {self.gender}
                Animal Type {self.Type}
                Animal food {self.food}
                """)

        
```

```python
class Dog(Animal):  # child class
    def __init__(self,name,age,color,breed,gender,Type,food, category,cattle):
        super().__init__(name,age,color,breed,gender,Type,food)
        self.category = category
        self.cattle = cattle
    def new_cartegory(self):
        print(f" Dog {self.category}")
    def new_cattle(self):
        print(f" Dog {self.cattle}")
    def walk(self):
        print("child class ka ")
        
```

```python
class Animal: # parent class
    def __init__(self,name,age,color,breed,gender,Type,food):
        self.name = name
        self.age = age
        self.color = color
        self.breed = breed
        self.gender= gender
        self.Type = Type
        self.food = food
    def sleep(self):
        print("Animal so skta hai ")
    def walk(self):
        print("parent class ka")
    def show_color(self):
        print(f" Animal color {self.color}")
    def food(self):
        print(f" animal food {self.food}")
    def full_info(self):
        print(f"""
                    Welcome
                Animal name {self.name}
                Animal age {self.age}
                Animal color {self.color}
                Animal breed {self.breed}
                Animal gender {self.gender}
                Animal Type {self.Type}
                Animal food {self.food}
                """)

        
```

```python
class Dog(Animal):  # child class
    def __init__(self,name,age,color,breed,gender,Type,food, category,cattle):
        super().__init__(name,age,color,breed,gender,Type,food)
        self.category = category
        self.cattle = cattle
    def new_cartegory(self):
        print(f" Dog {self.category}")
    def new_cattle(self):
        print(f" Dog {self.cattle}")
    def walk(self):
        print("child class ka ")
    def hello(self):
        super().walk()
    
        
```

```python
aa = Dog("bull",34,"Red","germanshafar","male","soft","meat","III","Farm")
```
