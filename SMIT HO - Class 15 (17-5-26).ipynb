```python
class Student:
    def __init__(self, name, roll, age, course):
        self.name = name
        self.roll = roll
        self.age = age
        self.course = course
        
    def read(self):
        print(f"{self.name} is reading..")
        
    def write(self):
        print(f"{self.name} is writing")
    
    def info(self):
        print(f"""
        name: {self.name}
        roll number: {self.roll}
        """)
```

```python
s1 = Student("usman", 1234, 20, "AI")
```

```python
class Car:
    def __init__(self, name, model, year, color, wheel, engine, interior):
        self.name = name
        self.model = model
        self.year = year
        self.color = color
        self.wheel = wheel
        self.engine = engine
        self.interior = interior
        
    def idle(self):
        print(f"{self.name} is idle")
        
    def start(self):
        print(f"{self.name} is starting")
    
    def stop(self):
        print(f"{self.name} is stop")
    
    def wheels(self):
        print(f"{self.name} have {self.wheel} wheels")
    
    def colors(self):
        print(f"the color is {self.color}")
        
    def engines(self):
        print(f" the engine is {self.engine}")

    def interiors(self):
        print(self.interior)
```

```python
car = Car("bmw", "m5cs", 2024, "black", 6, "v8", "leather")
```

```python
car.idle()
```

```python
car.colors()
```

```python
car.engines()
```

```python
car.color = "white"
```

```python
car.colors()
```

```python
class Car:
    def __init__(self, name, model, year, color, wheel, engine, interior):
        self.name = name
        self.model = model
        self.year = year
        self.color = color
        self.wheel = wheel
        self.engine = engine
        self.interior = interior
        
    def idle(self):
        print(f"{self.name} is idle")
        
    def start(self):
        print(f"{self.name} is starting")
    
    def stop(self):
        print(f"{self.name} is stop")
    
    def modify_color(self, cl):
        self.color = cl
        print(f"the color is changed your new color is {self.color}")
        
    def info(self):
        print(f"""
                    name: {self.name}
                    model: {self.model}
                    color: {self.color}
                    year: {self.year}
                    wheels: {self.wheel}
                    engine: {self.engine}
                    interior: {self.interior}
        
        
        """)
        
```

```python
car1 = Car("bmw", "m5cs", 2024, "black", 6, "v8", "leather")
```

```python
car1.modify_color("red")
```

```python
car1.color
```

```python
car1.info()
```
