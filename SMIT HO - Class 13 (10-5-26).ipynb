```python
def housing_cleaninig(room_list):
    for room in room_list:
        print("Checking :", room)
        if room =="Kitchen":
            print("kitchen ko clean rkho")
        elif room ==" bathroom":
            print("Bathroom ko wash kro ")
        elif room =="bedroom":
            print("Bedroom ko arrange karo")
        else:
            print(room, "Normal clean hai")
        print("Cleaning Complete\n")
```
