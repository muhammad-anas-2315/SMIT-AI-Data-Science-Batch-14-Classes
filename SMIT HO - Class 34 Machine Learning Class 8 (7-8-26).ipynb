```python
import pandas as pd
```

```python
data = pd.read_csv("/content/data.csv")
```

```python
data.isna().sum()
```

```python
data.dropna(inplace=True)
```

```python
data.isna().sum()
```

```python
data.duplicated().sum()
```

```python
data.select_dtypes("O")
```

```python
x= data.drop(["TenYearCHD"], axis= 1)
y= data.TenYearCHD
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test= train_test_split(x,y, test_size=0.25, random_state=42)
```

```python
X_train.shape, y_train.shape, X_test.shape, y_test.shape
```

```python
import warnings
warnings.filterwarnings("ignore")
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
y_train_pred = lr.predict(X_train)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
confusion_matrix(y_train, y_train_pred)
```

```python
accuracy_score(y_train, y_train_pred)
```

```python
model ne preict 1 kia  aur True hogia actual1 tha   ---- >   TP
model  ne predict 0 kia aur true hogia actual 0 tha   - ---->    TN
model nai preict 1 kia aur aur false bol kr actual mai false hopga ----- FP
model nai predict kia 0 aur false bol kr actual mai 1 tha      _____>    FN
```

```python
precison  =  predicted data pai kitne percent sahi hai
recall = actual data pai kitne percent sahi hai
```

```python
print(classification_report(y_train, y_train_pred))
```
